4. Citra yang dibaca dengan skimage.io.imread() ditampilkan berdampingan
dengan hasil cv2.cvtColor(img, cv2.COLOR_BGR2RGB) atas citra yang
sama. Manakah yang menampilkan warna sebenarnya, dan manakah yang
kanalnya tertukar? Buktikan dengan nilai piksel, bukan dengan pengamatan
visual saja.
- Warna Sebenarnya: Kedua cara tersebut sama-sama menghasilkan tampilan dengan warna asli yang benar. skimage.io.imread() secara native membaca citra dalam format RGB, sedangkan cv2.imread() yang dikonversi menggunakan cv2.cvtColor(..., cv2.COLOR_BGR2RGB) mengubah format BGR bawaan OpenCV menjadi RGB.
- Kanal yang Tertukar (sebagai perbandingan): Jika citra dibaca dengan cv.imread() tanpa konversi lalu langsung dianggap RGB, maka kanal warnanya tertukar (posisi nilai merah berada di tempat biru dan sebaliknya).

```python
# Membaca dengan skimage (Format RGB)
img_skimage = io.imread('ktp.png')

# Membaca dengan OpenCV, dan konversi ke RGB
img_cv_bgr = cv.imread('ktp.png')
img_cv_rgb = cv.cvtColor(img_cv_bgr, cv.COLOR_BGR2RGB)

# Membaca dengan OpenCV TANPA konversi
img_cv_raw = img_cv_bgr

# Membuktikan melalui sampel nilai piksel pada koordinat tertentu [0, 0]
print("Nilai piksel [0,0] (Format [R, G, B] pada skimage):", img_skimage[0, 0])
print("Nilai piksel [0,0] setelah cv2.cvtColor BGR -> RGB         :", img_cv_rgb[0, 0])
print("Nilai piksel [0,0] OpenCV mentah (Format [B, G, R]):", img_cv_raw[0, 0])
```

1. Tampilkan citra hanya pada kombinasi kanal Merah-Biru dan kanal HijauBiru saja, dengan kanal yang tidak dipakai dinolkan. Susun keduanya
berdampingan dalam satu figure yang diberi judul.

```python
img = cv.imread('ktp.png')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

img_mb = img_rgb.copy()  # Kombinasi Merah-Biru (Hijau/G = 0)
img_hb = img_rgb.copy()  # Kombinasi Hijau-Biru (Merah/R = 0)

img_mb[:, :, 1] = 0 # Menolkan kanal Hijau (indeks 1) pada citra Merah-Biru
img_hb[:, :, 0] = 0 # Menolkan kanal Merah (indeks 0) pada citra Hijau-Biru

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(img_mb)
axes[0].set_title("Kanal Merah-Biru (Hijau = 0)", fontsize=10)

axes[1].imshow(img_hb)
axes[1].set_title("Hijau-Biru (Merah = 0)", fontsize=10)

plt.show()
```

2. Tampilkan citra dalam tiga bentuk pencerminan: vertikal, horizontal, dan keduanya, dalam satu figure berisi tiga subplot yang masing-masing diberi label.

```python
img = cv.imread('ktp.png')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

flip_vertikal = cv.flip(img_rgb, 0)   #(Vertikal / Atas-Bawah)
flip_horizontal = cv.flip(img_rgb, 1) #(Horizontal / Kiri-Kanan)
flip_keduanya = cv.flip(img_rgb, -1)  #(Keduanya)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(flip_vertikal)
axes[0].set_title("Pencerminan Vertikal", fontsize=11)
axes[0].axis('off')

axes[1].imshow(flip_horizontal)
axes[1].set_title("Pencerminan Horizontal", fontsize=11)
axes[1].axis('off')

axes[2].imshow(flip_keduanya)
axes[2].set_title("Pencerminan Keduanya", fontsize=11)
axes[2].axis('off')

plt.show()
```

**Studi Kasus 1**: Panitia sebuah kegiatan perlu membubuhkan label identitas pada
ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel,
sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada
semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua
citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga
proporsinya tetap sama.

```python
import cv2 as cv
from google.colab.patches import cv2_imshow
import numpy as np

img = cv.imread('sariRoti.jpg')

if img is None:
    print("Error: Berkas gambar tidak ditemukan.")
else:
    h, w = img.shape[:2]
    bar_height = h // 12

    kanvas_label = img.copy()

    # Koordinat untuk bar label
    pt1_bar = (0, h - bar_height)
    pt2_bar = (w, h)

    cv.rectangle(kanvas_label, pt1=pt1_bar, pt2=pt2_bar, color=(0, 0, 0), thickness=-1)

    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = bar_height / 35.0
    thickness = max(1, int(bar_height / 15))

    org_text = (int(w * 0.03), h - int(bar_height * 0.3))

    # teks label
    cv.putText(kanvas_label,
               text='Dokumentasi',
               org=org_text,
               fontFace=font,
               fontScale=font_scale,
               color=(255, 255, 255),
               thickness=thickness,
               lineType=cv.LINE_AA)

    print(f"Ukuran Masukan: {w} x {h} px | Tinggi Bar Label: {bar_height} px")
    cv2_imshow(kanvas_label)
```

**Studi Kasus 2**: Marketplace seperti Tokopedia dan Shopee menampilkan foto
produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau
lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja.
Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail
persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna.
Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas,
tambahkan logo nama toko berukuran proporsional pada gambar.

```python
import cv2 as cv
from google.colab.patches import cv2_imshow
import numpy as np

img = cv.imread('sariRoti.jpg')

if img is None:
    print("Error: Berkas gambar tidak ditemukan.")
else:
    h, w = img.shape[:2]

    sisi = max(h, w)

    # Membuat kanvas persegi berwarna putih
    kanvas = np.full((sisi, sisi, 3), 255, dtype=np.uint8)

    # Menghitung koordinat agar gambar berada di tengah
    y_off = (sisi - h) // 2
    x_off = (sisi - w) // 2

    # Menempelkan citra asli ke tengah kanvas
    kanvas[y_off:y_off + h, x_off:x_off + w] = img

    # Menambahkan logo/nama toko secara proporsional
    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = sisi / 600.0  #
    thickness = max(1, int(sisi / 250))

    org_logo = (int(sisi * 0.05), int(sisi * 0.08))

    # Menambahkan teks nama toko
    cv.putText(kanvas,
               text='Toko Madura Ori',
               org=org_logo,
               fontFace=font,
               fontScale=font_scale,
               color=(0, 0, 255),
               thickness=thickness,
               lineType=cv.LINE_AA)

    print(f"Ukuran Asli: {w}x{h} px -> Diubah ke: {sisi}x{sisi} px")
    cv2_imshow(kanvas)
```
