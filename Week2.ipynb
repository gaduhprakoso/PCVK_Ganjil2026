3. Menampilkan citra Grayscale

```python
img_gray = cv.imread('ktp.png', cv.IMREAD_GRAYSCALE)

# Tanpa cmap
plt.subplot(1,2,1)
plt.imshow(img_gray)

# Dengan cmap
plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.show()
```

```python
#Jika ingin mengganti warna yang lain, tinggal mengganti cmap sesuai keinginan,
#contohnya untuk ditampilkan colormap dengan warna ‘magma
plt.imshow(img_gray, cmap='magma')
```

4. Melakukan resizing

```python
#mengubah ukuran citra dari ukuran 694 x 458 menjadi ukuran 347 x 229
imgRGB = cv.cvtColor(img, cv.COLOR_BGR2RGB)
img_output = cv.resize(imgRGB, (347, 229))
plt.imshow(img_output)
```

5. Melakukan Flipping

```python
#menampikan Citra RGB dalam posisi terbalik menggunakan fungsi flip
img_flip = cv.flip(img_output, 1)
plt.imshow(img_flip)
```

```python
#memperbesar ukuran canvas yang lebih besar dengan posisi gambar terbalik
imgFlip= cv.flip(img_output, 1)

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(1,1,1)
ax.imshow(imgFlip)
```

6. Membuat bentuk Geometri 2D dari OpenCV. Diawali dengan pembuatan
black image dengan tipe data int16.

```python
black_img = np.zeros((500, 530, 3), dtype=np.int16)
plt.imshow(black_img)
```

```python
#menambahkan bentuk persegi panjang sesuai koordinat pt1 dan pt2
cv.rectangle(black_img, pt1=(290, 0), pt2=(520, 130), color=(0, 0, 255), thickness=10)
plt.imshow(black_img)
```

```python
# menambahkan bentuk persegi sesuai koordinat pt1 dan pt2
cv.rectangle(black_img, pt1=(250, 250), pt2=(350, 350), color=(255, 0, 0), thickness=15)
plt.imshow(black_img)
```

```python
# menambahkan bentuk lingkaran
cv.circle(black_img, center=(150, 150), radius=60, color=(0, 255, 0), thickness=8)
plt.imshow(black_img)
```

```python
#penambahan lingkaran berwarna dan garis sesuai koordinat pt1 dan pt2
cv.circle(black_img, center=(400, 400), radius=50, color=(0, 255, 0), thickness=1)
cv.line(black_img, pt1=(0, 0), pt2=(530, 530), color=(255, 255, 255), thickness=5)
plt.imshow(black_img)
```

```python
#Penambahan text dengan font yang telah tertulis dengan ukuran yang sudah ditentukan.
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text='PCVK is fun!' , org=(10, 500), fontFace=font, fontScale=2,
           color=(255, 255, 0), thickness=3, lineType=cv.LINE_AA)
plt.imshow(black_img)
```

```python
#Pembuatan black image kembali dilakukan dengan tipe data int32
black_img2=np.zeros(shape=(500, 530, 3), dtype=np.int32)
plt.imshow(black_img2)
```

```python
#Berikut adalah kode program untuk inisialisasi NumPy array dengan tipe data int32
verticles = np.array([[150, 350], [250, 250], [450, 350], [250, 450]], dtype=np.int32)
verticles
```

```python
#Array tersebut kemudian di reshape sebagai berikut
pts = verticles.reshape(-1,1,2)
pts
```

```python
#Penambahan polyline pada black image kedua yang telah dibuat
cv.polylines(black_img2, [pts], isClosed=True, color=(0, 255, 0), thickness=5)
plt.imshow(black_img2)
```
